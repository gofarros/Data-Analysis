## Human Resource: Data-Driven for Better Retention 
Notebook: [Open Notebook](./HR Attrition/Notebook.ipynb)

This project focuses on the issue of attrition, which affects efficiency, productivity, and recruitment costs. As an HR Data Analyst, the author analyzes employee data to support data-driven decisions in recruitment, retention, and workforce strategy. The analysis includes visualizing hiring trends, total employees, workforce demographics, and factors such as tenure and travel frequency. The success of the analysis is measured using metrics like attrition rate, hiring trends, and correlations between key factors. Tools used include Python for data processing and Tableau Public for interactive visualization. The insights are used to optimize HR strategies through more accurate planning and policy-making.

- Skills: Data loading, cleaning, wrangling, exploratory data analysis, visualization.
- Technology/Tools: python, pandas, sqlalchemy, PostgreSQL, matplotlib, Tableau

## Boosting CTR and Conversion: Effective Online Campaign Strategies
Notebook: [Open Notebook](./Online Campaign/Notebook.ipynb)

This project evaluates past online campaigns to improve their effectiveness and efficiency. The goal is to gain insights from data to guide strategic decisions for future campaigns. Focus is on increasing key metrics like Click-Through Rate and Conversion Rate. Success will be measured by improvements in these conversion funnel metrics. Budget optimization and choosing the right promotional channels are key strategies. Regular evaluations will be conducted over three months to track progress and make adjustments. The project aims to achieve significant growth and help meet company revenue targets.

- Skills: Data loading, cleaning, wrangling, exploratory data analysis, hypotesis testing, visualization.
- Technology/Tools: python, pandas, scipy, matplotlib, Tableau

## Customer Churn: Insights for Internet Service Providers
Notebook: [Open Notebook](./ISP Customer Churn (SQL)/ISP Customer Churn.sql)

This project analyzes customer churn for an Internet Service Provider (ISP) to identify key factors influencing subscriber retention and attrition. Using data-driven techniques, the analysis examines customer behavior and the impact of several variables on churn. The goal is to uncover actionable insights that help reduce churn rates, improve customer loyalty, and optimize retention strategies. The project involves data extraction and analysis using SQL, supported by visualizations with Power BI to highlight trends and correlations. By understanding the root causes of churn, ISPs can make informed decisions to enhance service quality and customer satisfaction. The notebook contains SQL queries to process and analyze the data, enabling transparent and reproducible insights.

- Skills: Data loading, analysis, visualization.
- Technology/Tools: PostgreSQL, Power BI
